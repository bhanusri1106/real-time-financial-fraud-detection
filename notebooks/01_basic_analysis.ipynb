```python
#We will first train a Logistic Regression model because it is simple, fast to understand, and gives us a good baseline.
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    class_weight="balanced",
    random_state=42,
    max_iter=100
)

model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully")
```

```python
#make predictions on the test data.
```

```python
y_pred = model.predict(X_test_scaled)

print("Predictions completed successfully")
```

```python
import pandas as pd

pd.Series(y_pred).value_counts()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
#For fraud (1.0):

#Precision = 0.03 (3%) → Only about 3% of transactions predicted as fraud were actually fraud.

#Recall = 0.98 (98%) → The model detected about 98% of actual fraud transactions.

#F1-score = 0.05 → Low because precision is very low.

#For normal transactions (0.0):

#Precision = 1.00

#Recall = 0.95
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import roc_auc_score

y_prob = model.predict_proba(X_test_scaled)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", roc_auc)
```

```python
#ROC-AUC tells us how well the model can separate fraud transactions from normal transactions across different classification thresholds.
```

```python
#ROC-AUC is excellent, but precision is still only 3% at our current threshold
```

```python
#calculate PR-AUC (Average Precision), which is especially useful for our highly imbalanced fraud dataset.
```

```python
from sklearn.metrics import average_precision_score

pr_auc = average_precision_score(y_test, y_prob)

print("PR-AUC:", pr_auc)
```

```python
#The Logistic Regression baseline achieved a ROC-AUC of 0.995 and PR-AUC of 0.594. Since fraud is highly imbalanced, PR-AUC is important for evaluating the model’s ability to identify fraudulent transactions. The current model has high recall but low precision, so I will improve the precision-recall trade-off in the next stage.
```

```python
#Check a Better Prediction Threshold
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

threshold = 0.7

y_pred_07 = (y_prob >= threshold).astype(int)

print("Threshold:", threshold)
print("Precision:", precision_score(y_test, y_pred_07))
print("Recall:", recall_score(y_test, y_pred_07))
print("F1-Score:", f1_score(y_test, y_pred_07))
```

```python
import numpy as np

thresholds = np.arange(0.1, 1.0, 0.1)

for threshold in thresholds:
    y_pred_temp = (y_prob >= threshold).astype(int)
    precision = precision_score(y_test, y_pred_temp)
    recall = recall_score(y_test, y_pred_temp)
    f1 = f1_score(y_test, y_pred_temp)

    print(
        f"Threshold: {threshold:.1f} | "
        f"Precision: {precision:.3f} | "
        f"Recall: {recall:.3f} | "
        f"F1: {f1:.3f}"
    )
```

```python
thresholds = np.arange(0.01, 1.0, 0.01)

best_threshold = 0
best_f1 = 0

for threshold in thresholds:
    y_pred_temp = (y_prob >= threshold).astype(int)
    f1 = f1_score(y_test, y_pred_temp)

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold

print("Best Threshold:", best_threshold)
print("Best F1-Score:", best_f1)
```

```python
threshold = 0.99

y_pred_099 = (y_prob >= threshold).astype(int)

print("Threshold:", threshold)
print("Precision:", precision_score(y_test, y_pred_099))
print("Recall:", recall_score(y_test, y_pred_099))
print("F1-Score:", f1_score(y_test, y_pred_099))
```

```python
cm_099 = confusion_matrix(y_test, y_pred_099)

print(cm_099)
```

```python
print("Final Fraud Detection Metrics")
print("-----------------------------")
print("Precision:", precision_score(y_test, y_pred_099))
print("Recall:", recall_score(y_test, y_pred_099))
print("F1-Score:", f1_score(y_test, y_pred_099))
print("ROC-AUC:", roc_auc)
print("PR-AUC:", pr_auc)
```
